```python
# Create dataset
dataset = ["I love machine learning",
    "I love deep learning",
    "I enjoy coding",
    "I love NLP"]

# Tokenize data
token_list= []
for sentence in dataset:
    token_list.extend(sentence.lower().split())
unique_token_list= list(set(token_list))
print(f'{unique_token_list=}')

# Token ID
token_ids = {}
for idx, word in enumerate(unique_token_list):
    token_ids[word] = idx
print(f'{token_ids=}')

# Map dataset with token ids
dataset_mapped = []
for sentence in dataset:
    sentence_map= [token_ids[word] for word in sentence.lower().split()]
    dataset_mapped.append(sentence_map)
print(f'{dataset_mapped=}')

# Input-ouput creation
X_train = []
y_train = []
for data in dataset_mapped:
    print(data)
    for i in range(len(data)-1):
        X_train.append(data[i])
        y_train.append(data[i+1])
print(f'{X_train=}')
print(f'{y_train=}')
```
